```python
import torch
```

```python
original = torch.load('/data/jykang/research_models/antispoof/aasist/exp_result_evid_full/exp_result_evid_softplus/LA_AASIST_debug_ep100_bs24/weights/epoch_37_0.471.pth')
trained = torch.load('/data/jykang/research_models/antispoof/aasist/exp_result_ee/Freeze_LA_AASIST_ep100_bs24/weights/epoch_73_0.394.pth')
print(original['pool_hT2.proj.bias'])
print(trained['pool_hT2.proj.bias'])
```

```python
for key in original.keys():
    print(f"{key} weight diff: {torch.sum(original[key].data - trained[key].data)}")
```

```python
# BatchNorm과 관련된 키워드
bn_keywords = ['running_mean', 'running_var', 'num_batches_tracked']

# BatchNorm 값 대체
for key in trained.keys():
    if any(bn_key in key for bn_key in bn_keywords):
        if key in original:
            print(f"Replacing {key}")
            trained[key] = original[key]
        else:
            print(f"Key {key} not found in original model")

# 대체 완료된 가중치 저장
output_path = '/data/jykang/research_models/antispoof/aasist/exp_result_ee/epoch_73_bn_replaced.pth'
torch.save(trained, output_path)
print(f"Updated weights saved to {output_path}")
```

```python
trained_replaced = torch.load('/data/jykang/research_models/antispoof/aasist/exp_result_ee/epoch_73_bn_replaced.pth')
for key in original.keys():
    print(f"{key} weight diff: {torch.sum(original[key].data - trained_replaced[key].data)}")
```
